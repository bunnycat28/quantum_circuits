```python
import numpy as np
from qiskit import QuantumCircuit, transpile
from qiskit.visualization import plot_histogram

try:
    from qiskit_aer import AerSimulator
    backend = AerSimulator()
except ImportError:
    from qiskit.providers.basic_provider import BasicSimulator
    backend = BasicSimulator()

SHOTS = 100

# Gate colours: [fill, text]
circuit_style = {
    "displaycolor": {
        "x":       ["#003399", "#FFFFFF"],
        "h":       ["#CC3300", "#FFFFFF"],
        "measure": ["#A9A9A9", "#000000"],
    }
}

def run(qc):
    """Run a circuit for 100 shots and return the counts."""
    return backend.run(transpile(qc, backend), shots=SHOTS).result().get_counts()
```

## Predictions (before running)
The qubit always starts in |0⟩.

| Circuit | Gates | Prediction |
|---|---|---|
| **A** | X | X flips |0⟩ to |1⟩, so the result is **always 1** (100 out of 100). |
| **B** | X, X | The second X flips it back to |0⟩, so the result is **always 0** (100 out of 100). |
| **C** | H | H creates an equal superposition, so the result is **random, about 50% 0 and 50% 1**. |

**Will Circuit C always give the same result?** No. Each shot is random; only the overall distribution is about 50/50, and the exact counts will vary from run to run.

## Circuit A: X gate

```python
qc_A = QuantumCircuit(1, 1)      # qubit starts in |0>
qc_A.x(0)
qc_A.measure(0, 0)

qc_A.draw(output="mpl", style=circuit_style)
```

## Circuit B: two X gates

```python
qc_B = QuantumCircuit(1, 1)
qc_B.x(0)
qc_B.x(0)
qc_B.measure(0, 0)

qc_B.draw(output="mpl", style=circuit_style)
```

## Circuit C: H gate

```python
qc_C = QuantumCircuit(1, 1)
qc_C.h(0)
qc_C.measure(0, 0)

qc_C.draw(output="mpl", style=circuit_style)
```

## Run all three circuits (100 shots each)

```python
counts_A = run(qc_A)
counts_B = run(qc_B)
counts_C = run(qc_C)

print("Circuit A (X):    ", counts_A)
print("Circuit B (X, X): ", counts_B)
print("Circuit C (H):    ", counts_C)
```

## Histograms

```python
plot_histogram(counts_A, title="Circuit A: X gate",
               color="#003399", figsize=(7, 5))
```

```python
plot_histogram(counts_B, title="Circuit B: X, X gates",
               color="#6A7FB5", figsize=(7, 5))
```

```python
plot_histogram(counts_C, title="Circuit C: H gate",
               color="#CC3300", figsize=(7, 5))
```

## All three circuits together

```python
plot_histogram([counts_A, counts_B, counts_C],
               legend=["A: X", "B: X, X", "C: H"],
               color=["#003399", "#6A7FB5", "#CC3300"],
               title="Circuits A, B and C", figsize=(9, 5))
```

## Prediction vs actual

```python
predictions = {
    "A (X)":    {"0": 0,   "1": 100},
    "B (X, X)": {"0": 100, "1": 0},
    "C (H)":    {"0": 50,  "1": 50},
}
actual = {"A (X)": counts_A, "B (X, X)": counts_B, "C (H)": counts_C}

print(f"{'Circuit':<10} {'Pred 0':>7} {'Pred 1':>7} | {'Actual 0':>8} {'Actual 1':>8}")
for name in predictions:
    p, a = predictions[name], actual[name]
    print(f"{name:<10} {p['0']:>7} {p['1']:>7} | {a.get('0', 0):>8} {a.get('1', 0):>8}")
```

## Comparison and conclusion
- **Circuit A:** matches the prediction. The X gate flips |0⟩ to |1⟩, so every shot gives `1`.
- **Circuit B:** matches the prediction. Two X gates cancel out (X·X = identity), so every shot gives `0`.
- **Circuit C:** the H gate puts the qubit in superposition, so the results are random with roughly 50% `0` and 50% `1`. The exact split is not precisely 50/50 because only 100 shots are used, and it changes each time the circuit is run.

Circuits A and B are deterministic (the outcome is certain), while Circuit C is probabilistic.
